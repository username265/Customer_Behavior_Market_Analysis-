```python
df.isnull().sum()
#now weh have 37 duplicates values in col 'review rating'. Traditional step woudld to put the mean or median of the col in those places and move on.
#but considering the type of dataset we have which includes diff type of product/catergios and the rating may differ for each product and catergory.
#doing such way could skew our results and final analysis, instead to we replace it median of review rating with each category that way we can
# we can preserve the orginality and true nature of the dataset.
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
#feature engineering
```

```python
#1. create a col age_group
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
#qcut -> its splits the col and assigned the labels by itself
```

```python
df[['age','age_group']].head(10)
```

```python
#2. col purchase_frequency_days
#dictionary created
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```
